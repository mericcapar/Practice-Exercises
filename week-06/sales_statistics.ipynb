```python
import statistics
days = [
'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday',
'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'
]
sales = [1250, 1480, 1100, 1650, 1850,
1420, 1700, 1300, 2100, 1950]
```

```python
print('DAILY SALES REPORT')
print('-' * 18)
for i in range(len(days)):
    print(f"{days[i]} : ${sales[i]}")
```

```python
average = statistics.mean(sales)
mean = statistics.mode(sales)
std_dev = statistics.stdev(sales)
minimum = min(sales)
total = sum(sales)
maximum = max(sales)
median = statistics.median(sales)
mode = statistics.mode(sales)
print("DESCRIPTIVE STATISTICS")
print("-" * 22)
for i in range(len(days)):
    print(f"{days[i]} : ${sales[i]}")
print(f"Minimum: ${minimum:,.2f}")
print(f"Total: ${total:,.2f}")
print(f"TMaximum: ${maximum:,.2f}")
print(f"Standard deviaton: ${std_dev:,.2f}")
print(f"Median : ${median:,.2f}")
print(f"Mode : ${mode:,.2f}")
print(f"Average : ${average:,.2f}")
```

```python
performance = ['Excellent','Good','Average', 'Needs Improvement']
print('SALES PERFORMANCE')
print('-' * 18)
for i in range(len(days)):
    if sales[i] >= 1800:
        print(f"{days[i]} : ${sales[i]:,.2f} - {performance[0]}")
    elif sales[i] >= 1500:
        print(f"{days[i]} : ${sales[i]:,.2f} - {performance[1]}")
    elif sales[i] >= 1200:
        print(f"{days[i]} : ${sales[i]:,.2f} - {performance[2]}")
    else: 
        print(f"{days[i]} : ${sales[i]:,.2f} - {performance[3]}")
```
