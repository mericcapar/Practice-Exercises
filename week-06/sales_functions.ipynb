```python
employees = ['John', 'Sarah', 'Michael', 'Jennifer', 'David']
sales = [5200, 6800, 4500, 7200, 5800]
```

```python
def calculate_total(sales):
    total = 0
    for amount in sales:
        total += amount
    return total
```

```python
total = calculate_total(sales)
print(f"Total Monthly Sales: ${total:,.2f}")
```

```python
def calculate_average(total, number_of_employees):
    return total / number_of_employees
```

```python
average = calculate_average(total, len(employees))
print(f"Average Sales per Employee: ${average:,.2f}")
```

```python
def evaluate_performance(amount):
    if amount >= 7000:
        return "Excellent"
    elif amount >= 6000:
        return "Good"
    elif amount >= 5000:
        return "Average"
    else:
        return "Needs Improvement"
```

```python
print("EMPLOYEE SALES PERFORMANCE")
print("-" * 26)
for i in range(len(employees)):
    performance = evaluate_performance(sales[i])
    print(f"{employees[i]}: ${sales[i]:,.2f} - {performance}")
```

```python
def find_highest_sales(sales):
    highest = sales[0]
    for amount in sales:
        if amount > highest:
            highest = amount
    return highest
```

```python
highest = find_highest_sales(sales)
print(f"Highest Sales Amount: ${highest:,.2f}")
```

```python
def calculate_commission(amount, rate):
    return amount * rate
```

```python
commission = calculate_commission(5200, 0.05)
print(f"John's Commission: ${commission:,.2f}")
```

```python
print("EMPLOYEE SALES SUMMARY")
print("-" * 54)
print(f"{'Employee':<12}{'Sales':<12}{'Performance':<20}{'Commission'}")
print("-" * 54)

for i in range(len(employees)):
    performance = evaluate_performance(sales[i])
    commission = calculate_commission(sales[i], 0.05)
    sales_text = f"${sales[i]:,.2f}"
    commission_text = f"${commission:,.2f}"
    print(f"{employees[i]:<12}{sales_text:<12}{performance:<20}{commission_text}")
```

```python
def calculate_bonus_commission(amount):
    if amount >= 7000:
        rate = 0.10
    else:
        rate = 0.05
    return amount * rate

print(f"Jennifer's Commission: ${calculate_bonus_commission(7200):,.2f}")
print(f"John's Commission: ${calculate_bonus_commission(5200):,.2f}")
```

```python
def above_average(sales, average):
    count = 0
    for amount in sales:
        if amount > average:
            count += 1
    return count

above_count = above_average(sales, average)
print(f"Employees Above Average: {above_count}")
```

```python
def calculate_annual_sales(monthly_sales):
    return monthly_sales * 12

for i in range(len(employees)):
    annual = calculate_annual_sales(sales[i])
    print(f"{employees[i]}: ${annual:,.2f}")
```

1. A user-defined function is a function that I write myself using def, instead of a built-in one like print(). It groups a piece of code under a name so I can run it whenever I need it.

2. A parameter is the variable listed in the function definition, like amount and rate. An argument is the actual value I pass in when I call the function, like 5200 and 0.05.

3. The return statement sends a result from the function back to the place where it was called. This lets me store the result in a variable and use it later in the program.

4. print() only shows a value on the screen, while return sends the value back so the program can keep using it. A function that only prints gives back None, so I can't save or reuse its result.

5. Smaller functions make the program easier to read, test, and fix because each one does a single job. I can also reuse them, like how I used evaluate_performance() in both Task 3 and Task 6.

6. The function name is calculate_commission. The parameters are amount and rate. The arguments are 5200 and 0.05.
